# Dataset build

audit 결과를 사람이 확인한 뒤 canonical schema와 speaker split을 생성한다. 이 notebook에는 원본을 업로드하지 않는다.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd() / "src"))
from ulm.data.io import read_jsonl, write_jsonl
from ulm.data.split import split_by_speaker

records = read_jsonl("data/processed/canonical_unassigned.jsonl")
splits = split_by_speaker(records, seed=42)
for split, items in splits.items():
    write_jsonl(f"data/processed/{split}.jsonl", items, require_split=True)
```
