# 환경 확인

GPU와 Python 버전을 확인한다. 핵심 로직은 `src/ulm`에서 import한다.

```python
import platform
import sys

print(sys.version)
print(platform.platform())
try:
    import torch

    print("CUDA:", torch.cuda.is_available())
    if torch.cuda.is_available():
        print(torch.cuda.get_device_name(0))
except ImportError:
    print("torch가 아직 설치되지 않았습니다")
```
